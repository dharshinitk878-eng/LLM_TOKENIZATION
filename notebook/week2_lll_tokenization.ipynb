```python

import pandas as pd
from pathlib import Path

# Find your project folder
BASE_DIR = Path.cwd().parent
DATA_DIR = BASE_DIR / "dataset"

# Create the dataset folder if needed
DATA_DIR.mkdir(parents=True, exist_ok=True)

# Create practice text data
data = {
    "clean_text": [
        "Artificial intelligence helps computers understand human language.",
        "Large language models process text using tokens and numerical IDs.",
        "Python is useful for natural language processing and machine learning.",
        "BERT uses WordPiece tokenization to divide text into smaller units.",
        "GPT-2 uses byte pair encoding to represent text as tokens.",
        "Tokenization converts text into smaller pieces before model processing.",
        "Hugging Face provides pretrained tokenizers for language models.",
        "Different tokenizers may split the same sentence differently."
    ]
}

df = pd.DataFrame(data)

# Save the CSV file
csv_path = DATA_DIR / "clean_text.csv"
df.to_csv(csv_path, index=False)

print("Success! Your practice dataset has been created.")
print("File location:", csv_path)
print("Number of text records:", len(df))

display(df)
```

```python

import pandas as pd
import nltk
from pathlib import Path
from transformers import AutoTokenizer
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from IPython.display import display

print("Libraries imported successfully!")
```

```python

from transformers import AutoTokenizer

print("Loading BERT tokenizer...")
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
print("BERT loaded!")

print("Loading GPT-2 tokenizer...")
gpt2_tokenizer = AutoTokenizer.from_pretrained("gpt2")
print("GPT-2 loaded!")

print("Both tokenizers are ready!")
```

```python

# Compare BERT and GPT-2 tokens for every sentence

df["bert_tokens"] = df["clean_text"].apply(
    lambda text: bert_tokenizer.tokenize(str(text))
)

df["gpt2_tokens"] = df["clean_text"].apply(
    lambda text: gpt2_tokenizer.tokenize(str(text))
)

df["bert_token_count"] = df["bert_tokens"].apply(len)
df["gpt2_token_count"] = df["gpt2_tokens"].apply(len)

print("Tokenization completed successfully!")

display(
    df[[
        "clean_text",
        "bert_tokens",
        "gpt2_tokens",
        "bert_token_count",
        "gpt2_token_count"
    ]]
)
```

```python

import matplotlib.pyplot as plt

# Calculate total tokens for each tokenizer
bert_total = df["bert_token_count"].sum()
gpt2_total = df["gpt2_token_count"].sum()

print("Total sentences:", len(df))
print("Total BERT tokens:", bert_total)
print("Total GPT-2 tokens:", gpt2_total)

# Create a comparison chart
plt.figure(figsize=(7, 4))
plt.bar(
    ["BERT", "GPT-2"],
    [bert_total, gpt2_total]
)
plt.title("BERT vs GPT-2: Total Token Count")
plt.xlabel("Tokenizer")
plt.ylabel("Number of Tokens")
plt.show()

```


## Conclusion

In this experiment, I compared BERT and GPT-2 tokenization using a text dataset.

BERT uses WordPiece tokenization, while GPT-2 uses byte-level Byte Pair Encoding (BPE). Therefore, the two tokenizers may split the same text into different tokens and produce different token counts.

Tokenization is an important preprocessing step because language models convert text into tokens and numerical IDs before processing it.
